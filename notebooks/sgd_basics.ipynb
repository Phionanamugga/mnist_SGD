# ✅ Understanding the problem

# ✅ Import Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
```

# ✅ Define a Simple Quadratic Function (Loss Function)
# f(x) = x^2 + 3x + 2 (Example function to minimize)# 

```python
def function(x):
    return x**2 + 3*x + 2 
```

# ✅ Compute Gradient of the Function
# f'(x) = 2x + 3

```python
def gradient(x):
    return 2*x + 3
```

# ✅ Implement Stochastic Gradient Descent (SGD) from Scratch

```python
def stochastic_gradient_descent(learning_rate=0.1, epochs=20):
    x = np.random.uniform(-10, 10)
    x_values = [x] 
    y_values = [function(x)] 
    
    for i in range(epochs):
        grad = gradient(x)  
        x = x - learning_rate * grad  
        x_values.append(x)
        y_values.append(function(x))
        
        print(f"Epoch {i+1}: x = {x:.4f}, f(x) = {function(x):.4f}, Gradient = {grad:.4f}")
    
    return x_values, y_values
```

# ✅ Run SGD and Collect Results

```python
learning_rate = 0.1
epochs = 20
x_sgd, y_sgd = stochastic_gradient_descent(learning_rate, epochs)
```

# ✅ Compare with Batch Gradient Descent

```python
def batch_gradient_descent(learning_rate=0.1, epochs=20):
    x = np.random.uniform(-10, 10)
    x_values = [x]
    y_values = [function(x)]
    
    for i in range(epochs):
        grad = gradient(x) 
        x = x - learning_rate * grad
        x_values.append(x)
        y_values.append(function(x))
    
    return x_values, y_values

x_batch, y_batch = batch_gradient_descent(learning_rate, epochs)
```

# ✅ Visualization of SGD vs Batch GD

```python
plt.figure(figsize=(10, 5))
plt.plot(range(epochs+1), y_sgd, 'o-', label="SGD Progress", markersize=5)
plt.plot(range(epochs+1), y_batch, 's-', label="Batch GD Progress", markersize=5)
plt.xlabel("Epochs")
plt.ylabel("Function Value f(x)")
plt.title("SGD vs Batch Gradient Descent Optimization")
plt.legend()
plt.grid()
plt.show()
```
